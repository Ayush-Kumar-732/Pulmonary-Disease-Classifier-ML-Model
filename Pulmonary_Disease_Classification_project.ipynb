#### Chart - 6 Population Pyramid of Pulmonary Disease Patients by Age and Gender

#### Chart - 7 Average Energy Level Across Age Groups by Gender(Line Chart)

````python
# Chart -7 Visualization code
import pandas as pd
import matplotlib.pyplot as plt

# Create 5-year age groups
bins = [30, 35, 40, 45, 50, 55, 60, 65, 70, 75, 80, 85]
labels = ['30-34', '35-39', '40-44', '45-49', '50-54',
          '55-59', '60-64', '65-69', '70-74', '75-79', '80-84']

df['Age_Group'] = pd.cut(
    df['AGE'],
    bins=bins,
    labels=labels,
    right=False
)

# Calculate average energy level by age group
male = (
    df[df['GENDER'] == 1]
    .groupby('Age_Group', observed=True)['ENERGY_LEVEL']
    .mean()
    .reindex(labels)
)

female = (
    df[df['GENDER'] == 0]
    .groupby('Age_Group', observed=True)['ENERGY_LEVEL']
    .mean()
    .reindex(labels)
)

# Plot (No Rolling Average)
plt.figure(figsize=(10,6))

plt.plot(
    labels,
    male,
    color='royalblue',
    linewidth=2,
    marker='.',
    markersize=7,
    label='Male'
)

plt.plot(
    labels,
    female,
    color='mediumvioletred',
    linewidth=2,
    marker='.',
    markersize=6.5,
    label='Female'
)

plt.title("Average Energy Level Across Age Groups by Gender",
          fontsize=15,
          fontweight='bold')

plt.xlabel("Age Group (Years)")
plt.ylabel("Average Energy Level")

plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.xticks(rotation=45)

plt.legend()

plt.tight_layout()
plt.show()
````

#### Chart - 8 Age-wise Trend of Average Oxygen Saturation by Pulmonary Disease Status

````python
# Chart Visualization code
import matplotlib.pyplot as plt
import pandas as pd
from scipy.interpolate import make_interp_spline
import numpy as np

# Create age groups
bins = [30,35,40,45,50,55,60,65,70,75,80,85]
labels = ['30-34','35-39','40-44','45-49','50-54',
          '55-59','60-64','65-69','70-74','75-79','80-84']

df['Age_Group'] = pd.cut(
    df['AGE'],
    bins=bins,
    labels=labels,
    right=False
)

# Calculate average oxygen saturation
oxygen_data = (
    df.groupby(['Age_Group','PULMONARY_DISEASE'], observed=True)
    ['OXYGEN_SATURATION']
    .mean()
    .reset_index()
)

# Convert age groups to numerical positions
age_positions = np.arange(len(labels))

plt.figure(figsize=(10,6))

for disease, color, label in [
    (0, 'green', 'No Disease'),
    (1, 'red', 'Pulmonary Disease')
]:

    temp = oxygen_data[
        oxygen_data['PULMONARY_DISEASE'] == disease
    ]

    y_values = temp['OXYGEN_SATURATION'].values

    # Smooth curve
    x_smooth = np.linspace(
        age_positions.min(),
        age_positions.max(),
        300
    )

    spline = make_interp_spline(
        age_positions,
        y_values,
        k=3
    )

    y_smooth = spline(x_smooth)

    plt.plot(
        x_smooth,
        y_smooth,
        color=color,
        linewidth=2,
        label=label
    )

    # Add actual data points
    plt.scatter(
        age_positions,
        y_values,
        color=color,
        s=40
    )


plt.xticks(
    age_positions,
    labels,
    rotation=45
)

plt.title(
    "Oxygen Saturation Trajectory Across Age Groups by Pulmonary Disease Status",
    fontsize=15,
    fontweight='bold'
)

plt.xlabel("Age Group (Years)")
plt.ylabel("Average Oxygen Saturation (%)")

plt.grid(
    linestyle='--',
    alpha=0.4
)

plt.legend()

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A smooth trajectory plot was selected to visualize the changing pattern of average oxygen saturation across different age groups while comparing patients with and without pulmonary disease. The curved trend lines make it easier to observe variations and differences between the two groups and identify whether oxygen saturation patterns differ based on disease status.

##### 2. What is/are the insight(s) found from the chart?

* The average oxygen saturation levels remain relatively stable across all age groups, with values approximately between 94.8% and 95.2%.  
* Patients with pulmonary disease (red line) show slightly higher oxygen saturation values in several age groups, especially around 55–59, 70–74, and 80–84 years.  
* Patients without pulmonary disease (green line) show fluctuations, with noticeable decreases around 60–64 and 70–79 age groups.  
* The two trajectories overlap in multiple age groups, indicating that oxygen saturation alone may not clearly separate pulmonary disease and non-disease cases in this dataset.  
* This suggests that other factors such as smoking, pollution exposure, symptoms, and medical history should also be considered for accurate prediction.

##### 3. Will the gained insights help creating a positive business impact? Are there any insights that lead to negative growth? Justify with specific reason.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Define the continuous columns
continuous_cols = ['AGE', 'ENERGY_LEVEL', 'OXYGEN_SATURATION']

# Function to count outliers using the IQR method
def count_outliers(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Count rows where the value is outside the bounds
    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    return len(outliers)

print("Number of outliers detected (IQR Method):")
print("-" * 40)
for col in continuous_cols:
    outlier_count = count_outliers(df, col)
    print(f"{col}: {outlier_count} outliers")
````

````python
# Removing outliers using IQR(Interquartile Range) method
cols_to_clean = ['ENERGY_LEVEL', 'OXYGEN_SATURATION']

# Store original shape to see how many rows get removed
original_shape = df.shape

for col in cols_to_clean:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Reassign the dataframe to keep only the rows within the lower and upper bounds
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

    # Check outliers after removal
print("Number of outliers remaining (IQR Method):")
print("-" * 40)
for col in continuous_cols:
    outlier_count = count_outliers(df, col)
    print(f"{col}: {outlier_count} outliers")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?


**Technique Used:** Interquartile Range (IQR) Trimming.

I chose this method because it is highly robust and doesn't assume the data is normally distributed. Since the outliers in ENERGY_LEVEL and OXYGEN_SATURATION accounted for only ~1.2% of the dataset, trimming (removing) them entirely prevents the machine learning model from being biased by extreme values, with negligible data loss.
